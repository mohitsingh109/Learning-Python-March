```python
# ETL
# Merge & Concat

# merge() --> is used to combine two dataframes based on a common column or index.
#pd.merge(left, right, how='inner', on=None, left_on=None,right_on=None)
# left, right --> two dataframes to be merged
# on --> column name to merge on
# how --> type of merge to be performed (inner, outer, left, right)
# how='inner' --> only the rows with matching values in both dataframes will be included in the result
# left_on --> column name in left dataframe to merge on
# right_on --> column name in right dataframe to merge on
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5, 6],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Mohit', 'Karan'],
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 44],
    'Score': [25, 30, 35, 40],
})
df2
```

```python
# Inner Join
# Only row with matching key (on='ID') are included
result = pd.merge(df1, df2, on='ID', how='inner')
print(result)
```

```python
# Left Join
# All rows from the left df
result = pd.merge(df1, df2, on='ID', how='left')
result
```

```python
# Right Join
# All rows from the right df is included
result = pd.merge(df1, df2, on='ID', how='right')
result
```

```python
# Outer Join
# All rows from both df are included
# Outer = left + right
result = pd.merge(df1, df2, on='ID', how='outer')
result
```

```python
# Merge on multiple keys
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Subject': ['Science', 'Math', 'History'],
    'Score': [2, 4, 6]
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Subject': ['Math', 'History', 'English'],
    'Grade': ['A', 'B', 'C']
})
df2
```

```python
result = pd.merge(df1, df2, on=['ID', 'Subject'], how='inner')
result
```

```python
# Merge with different column names
df1 = pd.DataFrame({
    'Student_ID': [1, 2, 3, 1],
    'Subject': ['Science', 'Math', 'History', 'English'],
    'Score': [2, 4, 6, 7]
})
df1
```

```python
df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Subject': ['Math', 'History', 'English'],
    'Grade': ['A', 'B', 'C']
})
df2
```

```python
result = pd.merge(df1, df2, left_on=['Student_ID', 'Subject'], right_on=['ID', 'Subject'], how='inner')
result
```

```python
# Pandas Group By
# groupby() --> is used to group data based on one or more columns. It returns

# high --> 50 count, sum of cost column
# low --> 10
# medium --> 20
```

```python
data = {
    'Region': ['North', 'South', 'North', 'East', 'West', 'South', 'North'],
    'Product': ['A', 'A', 'C', 'A', 'B', 'C', 'A'],
    'Sales': [100, 200, 150, 100, 300, 400, 300],
    'Profit': [10, 20, 15, 5, 25, 30, 35]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Total sale for each region
grouped_sales = df.groupby('Region')['Sales'].sum()
print(grouped_sales)
```

```python
# Total sale for each region & product
grouped_sales = df.groupby(['Region', 'Product'])['Sales'].sum()
print(grouped_sales)
```

```python
# Total sale for each region & product
grouped_sales = df.groupby(['Region', 'Product'])['Sales'].agg(['sum', 'count', 'mean'])
print(grouped_sales)# Multiple Agg function
```

```python
# Rename agg column
grouped_sales = df.groupby(['Region', 'Product'])['Sales'].agg(total_sales='sum', avg_sales='mean')
print(grouped_sales)
```

```python
# Transform the groupby object to a DataFrame
# Why it skip the agg??
df['inr_amount'] = df.groupby('Region')['Profit'].transform(lambda dollar: dollar * 95)
df
```

```python
# Filter the groupby object to only include groups with a total sales greater than 400

filtered = df.groupby('Region').filter(lambda x: x['Sales'].sum() > 400)
print(filtered)
```

```python
# Group Keys & as_index
# as_index = True (group by key is your index)
grouped_sales = df.groupby('Region')['Sales'].sum()
grouped_sales
```

```python
# as_index = False (group by key is your column)
grouped_sales = df.groupby('Region', as_index=False)['Sales'].sum()
grouped_sales
```

```python
df_index = df.set_index('Region')
print(df_index.index)
df_index
```

```python
# Grouping by index
grouped = df_index.groupby(df_index.index)['Sales'].sum()
grouped
```

```python
# Group by custom function
grouped = df.groupby(lambda d: df.loc[d, 'Region'].startswith('N'))['Sales'].sum()
grouped
```

```python
# Agg on multiple columns

grouped = df.groupby('Region').agg({'Sales': 'sum', 'Profit': 'mean'})
grouped
```

```python
grouped = df.groupby('Region').agg({'Sales': ['sum', 'count'], 'Profit': ['sum', 'count']})
grouped
```
