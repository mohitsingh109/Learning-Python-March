```python
import matplotlib.pyplot as plt
```

```python
# 1.  Pyplot Approach - less code
X  = [1, 2, 3, 4, 5]
Y =  [2, 4, 6, 8, 10]

plt.plot(X, Y) # Matplotlib will automatically create a figure and axes for you

plt.title("Simple Line Plot")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
# 2 Object-Oriented Approach - more code but more control
# Encapsulation --> Getter/Setter to update or access the values

x  = [1, 2, 3, 4, 5]
y =  [2, 4, 6, 8, 10]

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Simple Line Plot")
ax.set_xlabel("X-axis") # to set the value we use setter
ax.set_ylabel("Y-axis")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
# Row = 2
# Col = 2
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100, 150, 200, 250, 300, 350]
profit = [20, 30, 40, 50, 60, 70]
orders = [10, 15, 20, 25, 30, 35]
customers = [5, 10, 15, 20, 25, 30]

# Sales Trend
axes[0, 0].plot(months, sales)
axes[0, 0].set_title('Sales Trend')

# Profit Trend
axes[0, 1].plot(months, profit)
axes[0, 1].set_title('Profit Trend')

# Orders Trend
axes[1, 0].plot(months, orders)
axes[1, 0].set_title('Order Trend')

# Customers Trend
axes[1, 1].plot(months, customers)
axes[1, 1].set_title('Customer Trend')

plt.tight_layout() # Automatically adjust spaces between chart

# Save the chart to png file
fig.savefig('company_data.png', dpi=300)
plt.show()
```

```python
# Create an empty figure
fig = plt.figure(figsize=(8, 5))
# width = 8 inches
# height = 5 inches
plt.show()
```

```python
# DPI: dots per inch
fig, ax = plt.subplots(
    figsize=(8, 5),
    dpi=100
)

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
# Change background color of figure
fig.set_facecolor('lightblue')

# Change background color of axes/chart
ax.set_facecolor('white')

ax.plot(x, y)
ax.set_title("DPI example")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
plt.savefig("example1.png", dpi=50)
plt.savefig("example2.png", dpi=100)
plt.savefig("example3.png", dpi=300)
plt.show()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(6, 6))

x = [1, 2, 3, 4, 5]
sales = [300, 150, 50, 250, 320]
profit = [41, 30, 40, 50, 35]

axes[0].plot(x, sales)
axes[0].set_title('Sales Trend')


axes[1].plot(x, profit)
axes[1].set_title('Profit Trend')

plt.tight_layout()
plt.show()
```
