```python
# Bar Plot

# category ==> Compare
# discrete value ==> Analyze

import matplotlib
print(matplotlib.__version__)
```

```python
import matplotlib.pyplot as plt
```

```python
# Creating bar chart
category = ['A', 'B', 'C', 'D', 'E']
values = [10, 20, 15, 25, 30]
plt.bar(category, values)
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.show()
```

```python
# Change bar chart color & align & width
plt.bar(category, values, color='red', width=0.3, align='center')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.show()
```

```python
# Add grid line (better readability)
plt.bar(category, values, color='green')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Bar Chart with Grid lines')

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Horizontal Bar chart
plt.barh(category, values, color='purple')
plt.xlabel('Values')
plt.ylabel('Category')
plt.title('Horizontal Bar Chart')

plt.show()
```

```python
# Add annotation on bar chart

plt.bar(category, values, color='green')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Bar Chart with annotations')

# adding annotation
for i, value in enumerate(values):
    plt.text(i, value + 0.5, str(value), ha='center', va='center')

plt.show()
```

```python
# Line Plot

numbers = [10, 20, 40, 40, 50, 40, 35]
plt.plot(numbers)
plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
temp = [30, 32, 35, 33, 31]

plt.plot(days, temp)
plt.title('Temperature during the week')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
# Add markers
# o --> Circle
# s --> square
# ^ --> triangle
# * --> Star
# + --> Plus
# D --> Diamond

plt.plot(days, temp, marker='D')
plt.title('Temperature during the week')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
# Change the line width

plt.plot(days, temp, linewidth=3, marker='o')
plt.title('Temperature during the week')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
# Change the line style

plt.plot(days, temp, marker='o', linestyle='--')
plt.title('Temperature during the week')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
# Change the line color

plt.plot(days, temp, marker='o', linestyle='--', color='red')
#plt.plot(days, [23, 26, 32, 43, 27], marker='o', linestyle='--', color='blue')
plt.title('Temperature during the week')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
